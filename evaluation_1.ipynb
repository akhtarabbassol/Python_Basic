### Question 1
Given a list of integers nums and an integer target, return <i>indices of the two numbers such that they add up to target<i>. <br>

Example 1:<br>

Input: nums = [2,7,11,15], target = 9<br>
Output: [0,1]

```python
nums = [2,7,11,15]
print(nums.index(2))
print(nums.index(7))
```

```python
nums = [2,7,11,15]
target = 9
res = []
for i in range(len(nums)):
    for j in range(i+1,len(nums)):
        if nums[i] + nums[j] == target:
            res.append(i)
            res.append(j)

print(res)

```

If the stack is full and someone tries to add an element, return a message saying the stack is full. If the stack is empty and someone tries to remove an element, return a message saying the stack is empty and no element can be removed.<br>

<b>Note:</b> Stack follows FIFO (first in first out) rule

```python
stack = []
max_size = 10
def pushing(n):
    if len(stack) == max_size:
        return "The Stack is Full"
    stack.append(n)
    return stack

def popping():
    if not stack:
        return "Stack is empty"
    stack.pop()
    return stack
for n in range(9): 
    pushing(n)

print(pushing(n))  
print(popping())




```

```python
l = [10,20,30,40,50]
def check_divisible(l):
    for i in l:
        if i%3==0 and i%5==0:
            print("fizzbuzz")
        elif i%5 == 0:
            print("buzz")
        elif i%3 == 0:
            print("fizz")
print(check_divisible(l))
```
